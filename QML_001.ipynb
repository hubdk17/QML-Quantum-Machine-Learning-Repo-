```python
import matplotlib.pyplot as plt
import numpy as np

u = np.array([1,0])

vectors = [
    np.array([1,0]),
    np.array([2,0]),
    np.array([0,1]),
    np.array([-1,0])
]

titles = [
    "Dot = 1",
    "Dot = 2",
    "Dot = 0",
    "Dot = -1"
]

plt.figure(figsize=(10,10))

for i,v in enumerate(vectors):

    plt.subplot(2,2,i+1)

    plt.quiver(0,0,u[0],u[1],
               angles='xy',
               scale_units='xy',
               scale=1,
               color='blue')

    plt.quiver(0,0,v[0],v[1],
               angles='xy',
               scale_units='xy',
               scale=1,
               color='red')

    plt.xlim(-2.5,2.5)
    plt.ylim(-2.5,2.5)

    plt.grid()

    plt.gca().set_aspect('equal')

    plt.title(titles[i])

plt.tight_layout()
plt.show()
```

Why do we use this formula for measuring similarity?

```python
import numpy as np

u = np.array([3,1])
v = np.array([1,2])

def f(u,v):
    return u[0]*v[0] + 2*u[1]*v[1]

print("Dot Product :", np.dot(u,v))
print("New Product :", f(u,v))
```

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 500)

# Standard unit circle
x1 = np.cos(theta)
y1 = np.sin(theta)

# Unit "circle" under weighted inner product
x2 = np.cos(theta)
y2 = np.sin(theta)/np.sqrt(2)

plt.figure(figsize=(7,7))

plt.plot(x1, y1, label="Standard Inner Product")
plt.plot(x2, y2, label="Weighted Inner Product")

plt.grid(True)
plt.axis("equal")

plt.legend()

plt.show()
```

It defines the geometry of the space.

Change the inner product...

↓

Change the notion of length.

↓

Change the notion of angle.

↓

Change the notion of distance.

In other words,

**The geometry depends on the inner product.**

```python
M = np.array([[1,0],
              [0,2]])

u = np.array([3,1])
v = np.array([1,2])

print(u.T @ M @ v)
```

We are NOT visualizing the dot product using circles.

We are visualizing the geometry induced by the inner product, and the circle naturally appears because of the definition of length.
**An inner product does not directly give us a circle.
It gives us a way to compute the length (norm) of a vector**

**Every valid quantum state lies on the unit sphere of the Hilbert space.**

```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([1.0, 0.0])

angles = np.arange(0, 361, 30)

plt.figure(figsize=(8,8))

for angle in angles:

    theta = np.radians(angle)

    v = np.array([
        np.cos(theta),
        np.sin(theta)
    ])

    dot = np.dot(u, v)

    plt.quiver(
        0,0,
        v[0],v[1],
        angles='xy',
        scale_units='xy',
        scale=1
    )

    plt.text(
        v[0]*1.1,
        v[1]*1.1,
        f"{angle}°\n{dot:.2f}",
        ha="center"
    )

plt.quiver(
    0,0,
    u[0],u[1],
    color="red",
    scale_units='xy',
    angles='xy',
    scale=1,
    label="u"
)
plt.axhline(0, color='black', linewidth=0.5) # x-axis
plt.axvline(0, color='black', linewidth=0.5) # y-axis
plt.axis('equal') # Ensures the circle looks round, not like an oval
plt.grid(True, linestyle='--')
plt.grid()
plt.axis("equal")
plt.xlim(-1.4,1.4)
plt.ylim(-1.4,1.4)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([1.0,0.0])

angles = np.linspace(0,360,500)

dots = []

for angle in angles:

    theta = np.radians(angle)

    v = np.array([
        np.cos(theta),
        np.sin(theta)
    ])

    dots.append(np.dot(u,v))

plt.figure(figsize=(9,4))

plt.plot(angles,dots)

plt.grid()

plt.xlabel("Angle")

plt.ylabel("Dot Product")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([4,0])      # horizontal vector
v = np.array([2,3])      # arbitrary vector

projection = np.array([v[0],0])

plt.figure(figsize=(7,7))

# u
plt.quiver(0,0,u[0],u[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='blue',
           label='u')

# v
plt.quiver(0,0,v[0],v[1],
           angles='xy',
           scale_units='xy',
           scale=1,
           color='red',
           label='v')

# projection line
plt.plot(
    [v[0],v[0]],
    [0,v[1]],
    '--',
    color='gray'
)

# projected vector
plt.quiver(
    0,0,
    projection[0],
    projection[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='green',
    label='Projection'
)

plt.grid()
plt.axis('equal')
plt.legend()

plt.xlim(-1,5)
plt.ylim(-1,4)

plt.show()
```
